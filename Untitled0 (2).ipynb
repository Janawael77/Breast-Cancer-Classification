```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()

x=pd.DataFrame(cancer.data ,columns=cancer.feature_names)
y=pd.DataFrame(cancer.target_names[cancer.target])

print(f"Number of patients(samples): {x.shape[0]}")
print(f"Number of Gene/cell features: {x.shape[1]}")
print("\nCancer Class Distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
print("Data successfully split and scaled!")
print(f"Training samples: {x_train.shape[0]}")
print(f"Testing samples: {x_test.shape[0]}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
model= RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(x_train_scaled , y_train)

y_pred= model.predict(x_test_scaled)
accuracy = accuracy_score(y_test , y_pred)

print(f"Model Accuracy: {accuracy *100:.2f}%\n")
print("Detailed Classification Report:")
print(classification_report(y_test , y_pred))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

cm= confusion_matrix(y_test , y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',
            xticklabels=['Benign','Malignant'],
            yticklabels=['Benign','Malignant'])
plt.xlabel('(Predicted Label)')
plt.ylabel('(True label)')
plt.title('Confusion Matrix - Breast Cancer Classification')
plt.show()
```

```python
import numpy as np
import joblib

sample_patient= x_test.iloc[0].values.reshape(1,-1)
sample_patient_scaled= scaler.transform(sample_patient)

prediction= model.predict(sample_patient_scaled)
probability= model.predict_proba(sample_patient_scaled)

actual_label= y_test.iloc[0]
predicted_label= prediction[0]

print("Single Patient prediction test ---")
print(f"Actual Diagnosis: {actual_label}")
print(f"Predicted Diagnosis: {predicted_label}")
print(f"Confidence Score: {np.max(probability) * 100:2f}%\n")

joblib.dump(model, 'breast_cancer_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
print("Model and Scaler successfully saved!")
```
